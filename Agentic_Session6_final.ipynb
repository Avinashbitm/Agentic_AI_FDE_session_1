Query Rewriting

```python
from langchain_core.documents import Document
from langchain_core.vectorstores import InMemoryVectorStore
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_text_splitters import CharacterTextSplitter, RecursiveCharacterTextSplitter

embeddings = HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")
```

```python
handbook = open("shopeasy_handbook.txt").read()

# Section-aware store from Part 1 (Strategy 3): one chunk per numbered section. The agent at the end searches it.
section_chunks = [
    Document(page_content=section, metadata={"section": section.split("\n")[0]})
    for section in handbook.split("\n\n")[1:]
]
section_store = InMemoryVectorStore.from_documents(section_chunks, embeddings)

print(len(section_chunks), "sections")
```

```python
def search(query):
    print(f"\nSearch: {query.strip()}")
    for doc in section_store.similarity_search(query, k=3):
        print(f"   - {doc.metadata['section']}")
```

```python
from langchain.tools import tool
from langchain.agents import create_agent

@tool
def search_handbook(query: str) -> str:
    """Search ShopEasy's customer policy handbook for returns, refunds, delivery, payments and contact details."""
    results = section_store.similarity_search(query, k=3)
    return "\n\n".join(doc.page_content for doc in results)
```

```python
# The agent mashed 3 questions into ONE search: "cash payment policy, return policy, refund timeline"
# Payment Methods never came back, so it answered "Yes, you can pay cash" -- WRONG (COD limit is Rs 10,000)
question = "Can I pay cash for a Rs 12,000 phone, return it if I don't like it, and how fast is the refund?"

agent = create_agent(
    model=llm,
    tools=[search_handbook],
    system_prompt="You are ShopEasy's customer support assistant. If you can't find the answer, say you don't know.",
)
response = agent.invoke({"messages": [{"role": "user", "content": question}]})
for message in response["messages"]:
    message.pretty_print()
```

```python
parts = llm.invoke(f"Split this question into separate questions, one per line:\n{question}").content

qa_pairs = ""
for part in parts.split("\n"):
    if part.strip():
        context = "\n\n".join(doc.page_content for doc in section_store.similarity_search(part, k=3))

        answer = llm.invoke([
            ("system", f"""You are ShopEasy's customer support assistant.
Answer only using the context below. If the answer is not there, say you don't know.

<context>
{context}
</context>"""),
            ("user", part),
        ]).content

        print(f"Q: {part}\nA: {answer}\n")
        qa_pairs += f"Question: {part}\nAnswer: {answer}\n\n"

final = llm.invoke(f"Here are some questions and answers:\n\n{qa_pairs}Use them to answer the customer's question in 3 to 4 short sentences, plain text: {question}").content
print("=== FINAL ANSWER ===")
print(final)
```
